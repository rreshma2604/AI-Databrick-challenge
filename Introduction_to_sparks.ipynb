# Creating a realistic dataset

```python

data = [
    (1, "view", "mobile", 29.99),
    (2, "view", "web", 15.49),
    (3, "cart", "mobile", 45.00),
    (4, "purchase", "web", 120.99),
    (5, "view", "mobile", 9.99)
]

columns = ["user_id", "event_type", "platform", "price"]

df = spark.createDataFrame(data, columns)
```

## View a sample of the data

```python
df.show()
```

## Checking the schema
Spark enforces structure. Correct schema = better performance + fewer bugs.

```python
df.printSchema()
```

## Comand to count rows

```python
row_count = df.count()
print(f"Total rows in dataset: {row_count}")

```

```python
from pyspark.sql.functions import count

df.groupBy("platform") \
  .agg(count("*").alias("event_count")) \
  .show()
```

### How Spark Handles Large Data

- Data is split into partitions
- Operations are executed in parallel
- Nothing runs until an action is called (lazy evaluation)
